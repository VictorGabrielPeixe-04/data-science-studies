Exercício 8

Crie um array com as vendas diárias [320, 150, 480, 290, 510, 180, 440] e imprima:

Melhor dia de vendas


Pior dia de vendas


Média diária

```python
diary_sale = np.array([320,450,480,390,510])

day_weeks = np.array(['Segunda-feira', 'Terça-feira', 'Quarta-feira', 'Quinta-feira', 'Sexta-feira'])
```

```python
for i in range(len(diary_sale)):
    print(f"Dia: {day_weeks[i]} : {diary_sale[i]}")
```

```python
#Melhor dia
better_day = np.argmax(diary_sale)
print(f"Melhor dia: {day_weeks[better_day]}\nValor do dia: {diary_sale[better_day]}")
```

```python
worst_day = np.argmin(diary_sale)
print(f"Pior dia: {day_weeks[worst_day]}\nTotal de venda: {diary_sale[worst_day]}")
```

```python
#Media diaria:
print(np.mean(diary_sale))
```

Crie um array com os preços [25.90, 49.90, 15.90, 89.90, 35.90, 199.90, 9.90]:

Ordene os preços

Imprima os 3 mais baratos

Calcule a média dos 3 mais baratos

```python
array_prices = np.array([25.90, 49,90, 15.90, 89.90, 35.90, 199.90, 9.90])
```

```python
order_prices = np.sort(array_prices)
print(order_prices)
```

```python
cheapest_prices = order_prices[:3]
print(cheapest_prices)
```

```python
mean_prices = np.mean(cheapest_prices)
print(f"A média dos preços foi de: {mean_prices:.2f}")
```

Exercício 10 — Desafio final

Crie três arrays:

products = ['Camiseta', 'Calça', 'Tênis', 'Boné']

sales = [150, 80, 60, 200]

price = [49.90, 129.90, 199.90, 39.90]

```python
products = np.array(['Camiseta', 'Calça', 'Tênis', 'Boné'])

sales = np.array([150, 80, 60, 200])

price = np.array([49.90, 129.90, 199.90, 39.90])
```

```python
for i in range(len(products)): 
    print(f"Produto {i + 1} : {products[i]}\n Peças vendidas:{sales[i]}\n Receita total: R${sales[i] * prices[i]}\n --------------------------")
```

```python
item_max = np.argmax(sales)
```

```python
print(f"Item mais vendido: {products[item_max]}\nTotal geral: R${np.dot(sales,price):.2f}")
```
