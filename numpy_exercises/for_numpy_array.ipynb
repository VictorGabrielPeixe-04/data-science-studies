Exercício 2

Crie um array com [15, 30, 45, 60, 75, 90, 105, 120], pegue os 4 primeiros, inverta e multiplique por 2.

```python
array_list = np.array([15, 30, 45, 60, 75, 90, 105, 120])
```

```python
#Pegando 4 primeiros numeros
get_numbers = array_list[:4]
print(get_numbers)
```

```python
#Invertendo a lista
print(get_numbers[::-1])

#Multiplicando os valores
print(get_numbers[::-1] * 2)
```

Exercício 3

Crie um array com ['React', 'Node', 'Python', 'Java', 'Go', 'Rust', 'C++'] e use for para imprimir cada linguagem com sua posição. Depois imprima somente as 3 últimas usando slice.

```python
array_lang = np.array(['React', 'Node', 'Python', 'Java', 'Go', 'Rust', 'C++'])
```

```python
for i in range(len(array_lang)):
    print(f"{i + 1}° ---- {array_lang[i]}")
```

```python
print(array_lang[4::])
```

```python
# Para deixar um pouco mais elegante, pode fazer desta maneira tambem:
print(array_lang[-3::])
```

Exercício 4

Crie um array com os números de 1 a 20 usando np.arange(). Sem usar for:

Pegue os números ímpares


Multiplique por 5

Some tudo com np.sum()

```python
array_num = np.arange(1, 21)
```

```python
#Array de 1 a 20
print(array_num)
```

```python
odd_numbers = array_num[::2]
print(odd_numbers)
```

```python
#Multiplicando
multi_numbers = (odd_numbers * 5)
print(multi_numbers)
```

```python
#Somando os numeros multiplicados
print(np.sum(multi_numbers))
```

Exercício 5

Crie um array com [100, 200, 300, 400, 500, 600, 700, 800, 900, 1000], inverta, pegue de 2 em 2 e some tudo.

```python
numeric_invert = np.array([100,200,300,400,500,600,700,800,900,1000])
```

```python
#Invertendo os valores
array_invert = numeric_invert[::-1]
print(array_invert)
```

```python
#Pegando de 2 em 2
get_numbers = array_invert[::2]
print(get_numbers)
```

```python
np.sum(get_numbers)
```
