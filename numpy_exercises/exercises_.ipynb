Exercício 3

Crie dois arrays:

products = ['Notebook', 'Mouse', 'Teclado', 'Monitor']
stock = [15, 200, 150, 30]

Use for para imprimir cada produto com seu estoque. No final imprima o produto com maior estoque e o com menor estoque.

```python
products = np.array(['Notebook', 'Mouse', 'Teclado', 'Monitor'])

stock = np.array([15, 200,150, 30 ])
```

```python
for i in range(len(products)):
    print(f"Produto: {products[i]}\nEstoque: {stock[i]}\n")
```

```python
#Produito com mais venda:
sales_stock = np.argmax(stock)
```

```python
#Maior venda
print(f"Produto: {products[sales_stock]}\nStock total: {stock[sales_stock]}")
```

```python
#Menor estoque
sales_stock_min = np.argmin(stock)
```

```python
print(f"Produto: {products[sales_stock_min]}\nEstoque total: {stock[sales_stock_min]}")
```

Crie um array com as temperaturas [36.5, 37.2, 38.1, 36.8, 39.0, 37.5, 36.2] e imprima:

Temperatura mais alta

Temperatura mais baixa

Média das temperaturas

Temperaturas ordenadas

```python
temperat = np.array([36.5, 17.2, 28.1, 26.8, 29.0, 27.5, 16.2])

day = np.array(['Segunda-feira', 'Terça-feira', 'Quarta-feira', 'Quinta-feira', 'Sexta-feira', 'Sabado,' ,'Domingo'])
```

```python
#Temperatura mais alta:
temp_high = np.argmax(temperat)
print(f"Dia mais quente: {day[temp_high]}\nTemperatura: {temperat[temp_high]}")
```

```python
#Temperatura mais baixa
temp_low = np.argmin(temperat)
```

```python
print(f"Dia mais frio: {day[temp_low]}\nTemperatura: {temperat[temp_low]}")
```

```python
#Média
print(f"Média das temperaturas: {np.mean(temperat):.2f}")
```

```python
#Temperaturas ordenadas
print(np.sort(temperat))
```

Crie um array com as notas [6.5, 8.0, 4.5, 9.5, 7.0, 3.5, 8.5, 5.0] e use for para imprimir cada nota com sua situação e no final imprima:

Maior nota

Menor nota

Média da turma

```python
grade_student = np.array([6.5, 8.0,  4.5, 9.5, 7.0 , 3.5, 8.5, 5.0])

student = np.array(['Victor', 'Cryscia', 'Kaue', 'Jose', 'Maria', 'Fernanda', 'Augusto', 'Gabriel'])
```

```python
max_grade = np.argmax(grade_student)
```

```python
print(f"Aluno com maior nota: {student[max_grade]}\nNota: {grade_student[max_grade]}")
```

```python
min_grade = np.argmin([grade_student])
```

```python
print(f"Aluno com a menor nota: {student[min_grade]}\nNota: {grade_student[min_grade]}")
```

```python
mean_class = np.mean(grade_student)
```

```python
print(f"A média da turma é de: {mean_class:.2f}")
```

```python
for i in range(len(grade_student)):
    if grade_student[i] >=6 and grade_student[i] <= 10:
        print(f"Nome: {student[i]}\nNota: {grade_student[i]}\nSituação: Aprovado\n")
    elif grade_student[i] == 5.75 and grade_student[i] < 6:
        print(f"Nome: : {student[i]}\nNota: {grade_student[i]}\nSituação: Avaliação final.\n")
    elif grade_student[i] < 5.74:
        print(f"Nome: {student[i]}\nNota: {grade_student[i]}\nSituação: reprovado\n")
```
