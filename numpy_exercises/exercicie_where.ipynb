- Exercício 7 — Combinando where + for

Crie um array com os nomes ['Victor', 'Maria', 'João', 'Ana', 'Pedro'] e um array com as idades [22, 17, 15, 25, 16]. Use np.where() para classificar como "Maior de idade" ou "Menor de idade" e use for para imprimir cada pessoa com sua classificação.

```python
names = np.array(['Victor', 'Maria', 'João', 'Ana', 'Pedro'])

year_old = np.array([22, 17, 15, 25, 16])
```

```python
for i in range(len(year_old)): 
    result = np.where(year_old[i] >= 18, "Maior de idade", "Menor de idade")
    print(f"O(a): {names[i]} tem {year_old[i]} anos. Então é : {result}")
```

- Continuação 7

```python
#O where pode ser fora tambem! 
attempt_other =  np.where(year_old >= 18, "Maior de idade", "Menor de idade")
```

```python
for j in range(len(year_old)): 
    print(f"O(a) {names[j]} tem {year_old[j]}. Então: {attempt_other[j]}")
```

Exercício 9 — Combinando where + argmax

- Crie um array com as vendas mensais [3200, 4100, 2800, 5000, 3700, 4500]. Use np.where() para classificar como "Bom mês" ou "Mês fraco". Depois imprima o melhor mês usando np.argmax().

```python
sales_month = np.array([3200, 4100, 2800, 5000, 3700, 4500])

month = np.array(['Janeiro', 'Fevereiro', 'Março', 'Abril', 'Maio', 'Junho'])
```

```python
#Média dos meses:
mean_month = np.mean(sales_month)
print(f"R$ {mean_month:.2f}")
```

```python
better_month = np.where(sales_month > mean_month, "Bom mês", "Mês fraco")
print(better_month)
print(sales_month)
```

```python
get_month = np.argmax(sales_month)
print(f"O melhor mês foi: {month[get_month]}\nReceita: {sales_month[get_month]}")
```

Exercício 10 — Desafio

Crie dois arrays:

students = ['Enrique', 'Enzo', 'João', 'Ana', 'Pedro', 'Lucas']
grades = [8.5, 5.0, 7.0, 4.5, 9.0, 6.0]

Use np.where() para classificar:

= 7 → "Aprovado"

= 5 e < 7 → "Recuperação"

< 5 → "Reprovado"

No final imprima:

Quantidade de aprovados

Quantidade em recuperação

Quantidade de reprovados

Média da turma

```python
students = np.array(['Enrique', 'Enzo', 'João', 'Ana', 'Pedro', 'Lucas'])

grade_student = np.array([8.5, 5.0, 7.0, 4.5, 9.0, 6.0])
```

```python
situation_stu = np.where(grade_student >=7 , "Aprovado",
                np.where((grade_student >= 5 ) & (grade_student < 7), "Recuperação", "Reprovado"))
```

```python
print(situation_stu)
```

```python
#Montando a quantidade de alunos aprovado

approved = np.sum(situation_stu == "Aprovado" )
print(f"Total de aprovados: {approved}")
```

```python
#Total de alunos para recuperaçã
recovery = np.sum(situation_stu == "Recuperação")
print(f"Total de alunos de recuperação: {recovery}")
```

```python
#Total de reprovados
disapproved = np.sum(situation_stu == "Reprovado")
print(f"Total de alunos reprovados: {disapproved}")
```

```python
#Média da turma.
mean_class = np.mean(grade_student)
print(f"Média da turma: {mean_class:.2f}")
```
