```python
array_month = np.array(['Janeiro',
                         'Fevereiro',
                         'Março',
                         'Abril', 
                         'Maio', 
                         'Junho', 
                         'Julho',
                         'Agosto', 
                         'Setembro', 
                         'Outubro', 
                         'Novembro', 
                         'Dezembro'])
```

Exercício extra 2 — For + indexação

Crie um array com [85, 92, 78, 95, 88, 73, 90] representando notas de alunos e use for para imprimir cada nota com sua situação. Exemplo:



Aluno 1: 85 - Aprovado


Aluno 2: 92 - Aprovado


Aluno 3: 78 - Reprovado

```python
grade_student = np.array([10, 4, 10, 5.50, 5, 0, 5.99, 6])
```

```python
for i in range(len(grade_student)):
    if grade_student[i] >= 6  and grade_student[i] <=10:
        print(f"Student {i + 1} : {grade_student[i]} \nSituation: Approved\n")
        
    elif grade_student[i] >5 and grade_student[i] <6:
        print(f"Student {i + 1} : {grade_student[i]}\nSituation: Second chance\n")
    
    else: 
        print(f"Student {i + 1} : {grade_student[i]}\nSituation: Disapproved\n")
```
