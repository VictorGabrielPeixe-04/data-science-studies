**Exercício 10 — Desafio**: 
Faça a limpeza completa do dataset e depois uma análise com groupby():

- Taxa de sobrevivência por classe
- Taxa de sobrevivência por sexo
- Idade média por classe
- Crie um gráfico de pizza com a distribuição de sobreviventes

```python
df = remove_column_
```

```python
df.head()
```

```python
with pd.option_context("float_format", "{:.2f}".format):
    display(df.describe())
```

```python
df.nunique()
```

```python
df["Pclass"].unique()
```

```python
df["Sex"].nunique()
```

```python
df["Sex"].value_counts()
```

```python
df.select_dtypes("number")
```

```python
df.isnull().sum()
```

```python
#Taxa de sobrevivência por classe
tax_survival = df.groupby("Pclass")["Survived"].mean()
color = plt.get_cmap('Dark2').colors
visualiztion = [0.03,0,0.1]
```

```python
plt.pie(tax_survival.values, labels=tax_survival.index, autopct="%1.2f%%", colors=color, explode=visualiztion)
plt.title("Taxa de sobrevivencia por classe")
plt.show()
```

```python
#Taxa de sobrevivência por sexo
tax_survived_sex = df.groupby("Sex")["Survived"].mean()
```

```python
plt.pie(tax_survived_sex.values, labels=tax_survived_sex.index, autopct="%1.2f%%")
plt.title("Taxa de sobrevivencia por sexo")
plt.xlabel("0: Male --- 1: Female")
plt.show()
```

```python
#Idade média por classe
age_mean_class = df.groupby("Pclass")[["Age"]].mean()
```

```python
with pd.option_context("float_format", "{:.0f}".format):
    display(age_mean_class)
```
