#### Erros cometidos

Exercício 6 — ⚠️ Usou df.describe().isnull().sum() — isso verifica nulos nas estatísticas, não no dataset!

  - Nesta primeira correção, pude perceber que cometi um erro. Describe() é usado para verificar as estatisticas em geral. Quando o Describe é utilizado, automaticamente ele já retira os valores nulos do dataframe. Por isso em meu caso deu errado. 

  - O certo é usar o df.isnull().sum() 

```python
df.isnull().sum()
```

Exercício 7 — ⚠️ Usou df.describe().nunique() — mesmo erro do 6! O correto é df.nunique()

  - Neste exercicio, o exercicio A está incorreto, pois usei o describe(). Como dito em cima, o describe é utilizado para gerar as estatisticas do nosso DF. 

  - O uso correto é utilizar somente a variavel DF. 

```python
df.nunique().sum()
```
